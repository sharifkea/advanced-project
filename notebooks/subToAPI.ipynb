```python
p2 ={"gender": 1,
  "height": 170,
  "ap_hi": 115,
  "ap_lo": 100,
  "cholesterol": 1,
  "gluc": 1,
  "smoke": 1,
  "alco": 1,
  "active": 0,
  "age_years": 45,
  "BMI": 25.5
}
p1 = []
for x in range(10):
    temp_dict = {}
    for y in p2:
        if y=="ap_hi" and x>5:
            temp_dict[y] = p2[y] + 10*(x%5)
        elif y=="ap_lo" and x<5:
            temp_dict[y] = p2[y] - 10*(x%5) 
        else:
            temp_dict[y] = p2[y]
    p1.append(temp_dict)
```

```python
import requests
import json
import pandas as pd

# API endpoint (update to your hosted URL)
API_URL = 'https://heartdiseaseprediction.pythonanywhere.com/predict'

def predict_heart_disease(input_data):
    """
    Function to make API call for heart disease prediction
    
    Parameters:
    input_data (dict): Dictionary containing input features for prediction
    
    Returns:
    dict: API response with prediction results
    """
    try:
        # Make POST request to the API
        response = requests.post(
            API_URL,
            headers={'Content-Type': 'application/json'},
            json=input_data  # Automatically converts to JSON
        )
        
        # Check if request was successful
        response.raise_for_status()
        
        # Parse JSON response
        result = response.json()
        
        return result
        
    except requests.exceptions.HTTPError as http_err:
        print(f'HTTP error occurred: {http_err}')
        return None
    except requests.exceptions.ConnectionError as conn_err:
        print(f'Connection error occurred: {conn_err}')
        return None
    except requests.exceptions.Timeout as timeout_err:
        print(f'Timeout error occurred: {timeout_err}')
        return None
    except requests.exceptions.RequestException as req_err:
        print(f'Request error occurred: {req_err}')
        return None
    except json.JSONDecodeError as json_err:
        print(f'JSON decode error: {json_err}')
        return None
```
